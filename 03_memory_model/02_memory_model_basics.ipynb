# 02 Memory Model Basics

Numpy handling memory:

Main focus:
- array ownership (`owndata`)
- view vs copy
- reshape and memory sharing
- how to verify with `np.shares_memory`

```python
import numpy as np
```

## 1) Array ownership

`owndata = True` means the array owns its memory buffer.
If it is `False`, the array may be a view of another array.

```python
arr = np.array([10, 20, 30, 40])
print("arr : ", arr)
print("arr.flags.owndata : ", arr.flags.owndata)
```

## 2) View from slicing

Slicing usually gives a **view**.
A view shares memory with the original array, so changing one can change the other.

```python
original = np.array([1, 2, 3, 4, 5])
view = original[1:4]

print("original : ", original)
print("view : ", view)
print("shares_memory(original, view) : ", np.shares_memory(original, view))
```

```python
view[0] = 99
print("After view[0] = 99")
print("original : ", original)
print("view : ", view)
```

## 3) Copy for independent memory

`copy()` creates a new array with separate memory.
So updates in copy do not affect the original array.

```python
a = np.array([5, 6, 7, 8])
b = a.copy()

print("a : ", a)
print("b : ", b)
print("shares_memory(a, b) : ", np.shares_memory(a, b))
```

```python
b[0] = -1
print("After b[0] = -1")
print("a : ", a)
print("b : ", b)
```

## 4) `reshape` often shares memory

`reshape` often returns a view (when memory layout allows).
That is why changing reshaped array may also change original array.

```python
x = np.arange(6)
y = x.reshape(2, 3)

print("x : ", x)
print("y : ", y)
print("shares_memory(x, y) : ", np.shares_memory(x, y))
```

```python
y[0, 0] = 100
print("After y[0,0] = 100")
print("x : ", x)
print("y : ", y)
```

## Quick points:

- Slicing usually returns a view (shared memory).
- `copy()` returns independent memory.
- `reshape()` often returns a view.
- Use `np.shares_memory(a, b)` to confirm whether arrays share memory.
