# 04 Ravel vs Flatten

Difference between ravel() and flatten():

- ravel() often gives a view (can share memory)
- flatten() always gives a copy (independent memory)

```python
import numpy as np
```

## 1) Create a 2D array

```python
a = np.array([[1, 2, 3], [4, 5, 6]])
print("a :", a)
```

## 2) Use ravel()

ravel() returns a flattened array.
In many cases it shares memory with the original array.

```python
r = a.ravel()
print("r : ", r)
print("np.shares_memory(a, r) : ", np.shares_memory(a, r))
```

```python
r[0] = 99
print("After r[0] = 99")
print("a :", a)
print("r : ", r)
```

## 3) Use flatten()

flatten() also returns a 1D array, but it always creates a copy.
So changes in flattened array do not affect the original.

```python
f = a.flatten()
print("f : ", f)
print("np.shares_memory(a, f) : ", np.shares_memory(a, f))
```

```python
f[1] = -1
print("After f[1] = -1")
print("a :", a)
print("f : ", f)
```

## Quick recap

- ravel() -> often view -> may share memory
- flatten() -> copy -> does not share memory
