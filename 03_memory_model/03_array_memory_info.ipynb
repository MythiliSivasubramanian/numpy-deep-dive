# 03 Array Memory Info

This notebook explains how to inspect memory-related properties of NumPy arrays in a simple way.

We will cover:
- `dtype`
- `itemsize`
- `size` and `nbytes`
- `base`
- `np.shares_memory()`

```python
import numpy as np
```

## 1) Create an array and inspect basic memory info

`dtype` tells data type.
`itemsize` tells bytes per element.
`size` tells number of elements.
`nbytes` tells total bytes used by elements.

```python
a = np.array([10, 20, 30, 40], dtype=np.int32)
print("a : ", a)
print("a.dtype : ", a.dtype)
print("a.itemsize (bytes per element) : ", a.itemsize)
print("a.size (number of elements) : ", a.size)
print("a.nbytes (total bytes) : ", a.nbytes)
print("a.ndim : ", a.ndim)
print("a.shape : ", a.shape)
```

## 2) View using slicing and `base`

Slicing often returns a view.
If array `b` is a view of `a`, then `b.base` usually points to `a` (or original owner).
`np.shares_memory(a, b)` helps confirm memory sharing.

```python
b = a[1:4]
print("b : ", b)
print("b.base is a : ", b.base is a)
print("np.shares_memory(a, b) : ", np.shares_memory(a, b))
```

```python
b[0] = 999
print("After b[0] = 999")
print("a : ", a)
print("b : ", b)
```

## 3) Copy and independent memory

`copy()` creates a new independent memory block.
So `np.shares_memory(a, c)` becomes `False`.
Also `c.base` is usually `None` because `c` owns its data.

```python
c = a.copy()
print("c : ", c)
print("c.base : ", c.base)
print("np.shares_memory(a, c) : ", np.shares_memory(a, c))
```

```python
c[0] = -1
print("After c[0] = -1")
print("a : ", a)
print("c : ", c)
```

## Quick recap

- `itemsize` = bytes used by one element
- `nbytes` = total bytes used by all elements
- `base` helps identify if array is a view
- `np.shares_memory(a, b)` checks shared memory between arrays
