# NumPy Indexing Pipelines: Filter → Pick vs Pick → Filter

This notebook explains **step‑by‑step, deeply visual way** with multiple examples.

Lets focus now three powerful indexing pipelines:

1. **Filter → then pick positions**
2. **Pick positions → then filter**
3. **Stable / safe pipeline (with size checks)**

Understanding the order of operations here is critical for real data work.


## 1) Setup — Create the array

````python
import numpy as np

a = np.array([15, 3, 22, 8, 7, 30, 11, 2, 18])
a
````

## 2) Two core indexing tools

Before scenarios, remember:

### A) Boolean filtering
```
a[a > 10]
```

### B) Positional picking
```
a[[0, 2, 5]]
```

Pipelines combine both — order matters a LOT.


````python
# Boolean filtering
a[a > 10]
````

````python
# Positional picking
a[[0, 2, 5]]
````

## 3) Scenario 1 — Filter → Then Pick

From code:

```python
filtered = a[a > 3]
result = filtered[[0, 2]]
```

### Step logic
1. Remove values ≤ 3
2. From the remaining array, pick positions 0 and 2


````python
filtered = a[a > 3]
result = filtered[[0, 2]]

print("a:", a)
print("filtered (>3):", filtered)
print("pick [0,2]:", result)
````

### Visualizing index reset

After filtering, indices restart from 0.


````python
list(enumerate(filtered))
````

So `[0,2]` refers to positions inside the **filtered array**, not the original.


## 4) Scenario 2 — Pick Positions → Then Filter

From code:

```python
picked = a[[0, 2, 3, 6]]
result = picked[picked % 2 == 0]
```

### Step logic
1. Extract specific positions
2. Keep only even numbers from that subset


````python
picked = a[[0, 2, 3, 6]]
result = picked[picked % 2 == 0]

print("a:", a)
print("picked [0,2,3,6]:", picked)
print("even only:", result)
````

### Why order matters:

Compare:

- Filter → Pick
- Pick → Filter

They can produce completely different outputs.


````python
# Compare both pipelines

pipeline1 = a[a > 3][[0, 2]]
pipeline2 = a[[0, 2, 3, 6]][a[[0, 2, 3, 6]] % 2 == 0]

print("Filter → Pick:", pipeline1)
print("Pick → Filter:", pipeline2)
````

## 5) Scenario 3 — Stable / Safe Pipeline

From code:

```python
filtered = a[(a >= 5) & (a <= 20)]

if filtered.size >= 3:
    result = filtered[[0, 1, -1]]
else:
    result = filtered
```

### Why this matters

If we try to pick indices that don’t exist → error.

So we check `.size` first.


````python
filtered = a[(a >= 5) & (a <= 20)]

if filtered.size >= 3:
    result = filtered[[0, 1, -1]]
else:
    result = filtered

print("a:", a)
print("filtered (5..20):", filtered)
print("safe result:", result)
````

### Another example: 


````python
b = np.array([1, 2, 3, 4, 5])

filtered_b = b[(b >= 5) & (b <= 20)]

if filtered_b.size >= 3:
    result_b = filtered_b[[0, 1, -1]]
else:
    result_b = filtered_b

print("b:", b)
print("filtered:", filtered_b)
print("safe result:", result_b)
````

## 6) Original Functions 


````python
def scenario_filter_then_pick(a: np.ndarray) -> None:
    filtered = a[a > 3]
    result = filtered[[0, 2]]
    print("Scenario: filter > 3 then pick [0,2]")
    print("a:", a)
    print("filtered:", filtered)
    print("result:", result)
    print()


def scenario_pick_positions_then_filter(a: np.ndarray) -> None:
    picked = a[[0, 2, 3, 6]]
    result = picked[picked % 2 == 0]
    print("Scenario: pick [0,2,3,6] then keep even")
    print("a:", a)
    print("picked:", picked)
    print("result:", result)
    print()


def scenario_stable_pipeline(a: np.ndarray) -> None:
    filtered = a[(a >= 5) & (a <= 20)]
    if filtered.size >= 3:
        result = filtered[[0, 1, -1]]
    else:
        result = filtered
    print("Scenario: safe pipeline")
    print("a:", a)
    print("filtered:", filtered)
    print("result:", result)
    print()


a = np.array([15, 3, 22, 8, 7, 30, 11, 2, 18])

scenario_filter_then_pick(a)
scenario_pick_positions_then_filter(a)
scenario_stable_pipeline(a)
````
