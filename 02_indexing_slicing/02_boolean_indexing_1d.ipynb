# Boolean Masking Scenarios — Thresholds, Properties, Replacement

This notebook demonstrates practical **Boolean Masking scenarios** in NumPy.

- Threshold filtering (`>`, `<`)
- Property filtering (even / odd)
- Conditional replacement using masks
- Safe copying before modification

These are foundational techniques for **data cleaning and preprocessing**.


````python
import numpy as np
````

## 1️⃣ Scenario: Threshold Filtering

We filter values based on numeric thresholds:

- Values greater than 4
- Values less than 5

Boolean masks return only the elements that satisfy the condition.


````python
a = np.array([3, 8, 2, 9, 5, 1, 12, 7])

gt_4 = a[a > 4]
lt_5 = a[a < 5]

print("a:", a)
print("> 4:", gt_4)
print("< 5:", lt_5)
````

## 2️⃣ Scenario: Property Filtering (Even / Odd)

We classify numbers using modulo `%`:

- Even → `a % 2 == 0`
- Odd → `a % 2 != 0`


````python
a2 = np.array([10, 11, 12, 13, 14, 15, 16])

evens = a2[a2 % 2 == 0]
odds = a2[a2 % 2 != 0]

print("a2:", a2)
print("even:", evens)
print("odd :", odds)
````

## 3️⃣ Scenario: Replacement with Boolean Mask

We modify values conditionally:

Rules:

- Replace values `< 0` → 0
- Replace values `> 10` → 10

We first create a copy to avoid modifying the original array.


````python
a3 = np.array([-5, 2, 7, 11, 25, 0, -1])

b = a3.copy()

b[b < 0] = 0
b[b > 10] = 10

print("Original a3:", a3)
print("Modified b :", b)
````

## Why `.copy()` Matters

If we modify the original array directly:

```python
a3[a3 < 0] = 0
```

We lose the original data.

Using `.copy()` ensures safe data preprocessing workflows.


## Practice Exercises

Try these on your own:

1. Filter values greater than 10.
2. Extract multiples of 3.
3. Replace values > 50 with 50.
4. Replace negative values with the array mean.

Solutions below 👇


````python
arr = np.array([5, 12, 18, -3, 7, 60, 45])

# 1
print("Greater than 10:", arr[arr > 10])

# 2
print("Multiples of 3:", arr[arr % 3 == 0])

# 3
capped = arr.copy()
capped[capped > 50] = 50
print("Capped at 50:", capped)

# 4
mean_val = arr.mean()
replaced = arr.copy()
replaced[replaced < 0] = mean_val
print("Negatives replaced with mean:", replaced)
````
