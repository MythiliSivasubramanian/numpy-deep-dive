# NumPy Fancy Indexing (Picking Positions): Basic Picks, Order, Duplicates, and Assignment

Fancy indexing means selecting elements using an **array/list of indices**, like:

```python
a[[0, 3, 4]]
```

- Basic positional picks
- Re-ordering results by changing index order
- Duplicates (repeating indices repeats values)
- Assigning values using fancy indices (with a safe copy)
- Common mistakes and best practices

## 3) Scenario 1 — Basic picks `[0,3,4]`

From code:

```python
picks = a[[0, 3, 4]]
```

### What happens?
- `0` selects the **first** element
- `3` selects the **4th** element
- `4` selects the **5th** element

The result is a **new array** with those selected values.


````python
picks = a[[0, 3, 4]]
print("a:", a)
print("picks [0,3,4]:", picks)
````

### Visualize positions
A helpful trick is to see `(index, value)` pairs.


````python
list(enumerate(a))
````

## 4) Scenario 2 — Order change + duplicates

From code:

```python
order_change = a[[4, 0, 3]]
duplicates = a[[1, 1, 3, 3]]
```

### Key rules
- **The output order follows your index list**.
- If you repeat indices, you repeat values.


````python
order_change = a[[4, 0, 3]]
duplicates = a[[1, 1, 3, 3]]

print("a:", a)
print("order [4,0,3]:", order_change)
print("duplicates [1,1,3,3]:", duplicates)
````

### More examples (practice)

- Reverse the array using indices
- Pick the last, then first, then last again
- Pick every other index using a list


````python
# Reverse using fancy indices
rev_idx = np.arange(a.size - 1, -1, -1)
rev = a[rev_idx]

# last, first, last again
lfla = a[[-1, 0, -1]]

# every other index (0,2,4,6,...)
every_other_idx = np.arange(0, a.size, 2)
every_other = a[every_other_idx]

print("reverse idx:", rev_idx)
print("reverse:", rev)
print("[-1,0,-1]:", lfla)
print("every other idx:", every_other_idx)
print("every other:", every_other)
````

## 5) Scenario 3 — Assignment with fancy indices

From code:

```python
b = a.copy()
idx = np.array([1, 3, 5])
b[idx] = -1
```

### Why `copy()`?
Because we usually want to **keep the original array `a` unchanged**.

### What does `b[idx] = -1` mean?
"At positions 1, 3, and 5, replace the values with `-1`."


````python
b = a.copy()
idx = np.array([1, 3, 5])
b[idx] = -1

print("a:", a)
print("idx:", idx)
print("b after assignment:", b)
````

### Assignment with duplicates :

If we assign with duplicate indices, the same position is assigned multiple times.
With a single scalar value like `-1`, that’s fine.

But with different values, duplicate indices can be tricky.


````python
b2 = a.copy()
idx2 = np.array([1, 1, 3])

# Scalar assignment (safe)
b2[idx2] = 99
print("Scalar assignment with duplicates:", b2)

# Value assignment with duplicates
b3 = a.copy()
vals = np.array([10, 20, 30])
b3[idx2] = vals
print("Value assignment with duplicates:", b3)
print("Note: index 1 was assigned twice; the later write wins for that position.")
````

## 6) Some common mistakes & best practices

### Mistake 1: Confusing values with indices
Fancy indexing expects **positions**, not values.
If we do `a[[3, 8]]`, NumPy treats `3` and `8` as *indices* — and `8` is out of range here.

### Mistake 2: Indices must be integers
Floats won't work:
```python
a[[1.0, 2.0]]
```

### Best practice: validate indices
- Use `np.max(idx) < a.size`
- Use `np.min(idx) >= -a.size` (if we allow negative indices)


````python
# Example: validate indices safely
idx = np.array([1, 3, 5])

print("idx:", idx)
print("max(idx) < a.size:", idx.max() < a.size)
print("min(idx) >= -a.size:", idx.min() >= -a.size)
````

## 7) Code from .py file



````python
import numpy as np


def scenario_basic_picks(a: np.ndarray) -> None:
    picks = a[[0, 3, 4]]
    print("Scenario: basic picks [0,3,4]")
    print("a:", a)
    print("picks:", picks)
    print()


def scenario_order_and_duplicates(a: np.ndarray) -> None:
    order_change = a[[4, 0, 3]]
    duplicates = a[[1, 1, 3, 3]]
    print("Scenario: order + duplicates")
    print("a:", a)
    print("order [4,0,3]:", order_change)
    print("duplicates [1,1,3,3]:", duplicates)
    print()


def scenario_assign_with_fancy(a: np.ndarray) -> None:
    b = a.copy()
    idx = np.array([1, 3, 5])
    b[idx] = -1
    print("Scenario: assign with fancy indices")
    print("a:", a)
    print("idx:", idx)
    print("b:", b)
    print()


def main() -> None:
    a = np.array([3, 8, 2, 9, 5, 1, 12])
    scenario_basic_picks(a)
    scenario_order_and_duplicates(a)
    scenario_assign_with_fancy(a)


main()
````
