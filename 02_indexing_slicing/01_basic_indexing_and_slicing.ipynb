```python
import numpy as np
```

NumPy is imported as np. All array operations in this notebook use NumPy arrays (ndarray).

```python
# Create a 2D array
x = np.arange(12).reshape(3, 4)
print(x)
print("shape:", x.shape, "ndim:", x.ndim)
```

We create numbers 0 to 11 using np.arange(12) (total 12 elements).
Then we reshape into a 3×4 table:

shape = (3, 4) means 3 rows and 4 columns

ndim = 2 because it has two axes:

axis 0 → rows

axis 1 → columns

So we can locate any element using two indices: [row, column].

```python
# Basic indexing (single element)
print("x[0, 0] =", x[0, 0])
print("x[2, 3] =", x[2, 3])
```

Basic indexing selects a single element.

x[0, 0] means: row 0, column 0 (top-left)

x[2, 3] means: row 2, column 3 (third row, fourth column)

Indexing starts at 0, not 1.

So for shape (3, 4):

valid row indices: 0, 1, 2

valid column indices: 0, 1, 2, 3

```python
# Slicing (rows/cols ranges)
print("\nFirst row:", x[0, :])
print("\nFirst column:\n", x[:, 0])
print("\nSubmatrix (rows 0-1, cols 1-2):\n", x[0:2, 1:3])
```

Slicing selects a range of elements.

1) x[0, :]

0 selects the first row

: means all columns
Result: a row vector with shape (4,)

2) x[:, 0]

: means all rows

0 selects the first column
Result: a column vector with shape (3,)

3) x[0:2, 1:3]

This is a submatrix (a smaller table from the big table).

0:2 → rows 0 and 1 (2 is not included)

1:3 → columns 1 and 2 (3 is not included)

So output shape becomes (2, 2).

⚡ Slicing uses half-open ranges: start included, stop excluded.

```python
# Slicing is usually a VIEW (prove it)
sub = x[:, 1:3]
sub[0, 0] = 999
print("sub:\n", sub)
print("x changed too:\n", x)
```

This is the most important slicing behavior:

Basic slicing usually creates a view (not a copy).

A view means:
    1. sub does not store new data
    2. it “looks at” the same memory as x
    3. therefore changing sub changes x

Here:
    sub = x[:, 1:3] selects all rows, columns 1 and 2
    When we set sub[0, 0] = 999, we are editing the original memory
    So x changes as well

This is why slicing is memory efficient — but can cause bugs if you forget it shares data.

```python
# Reset for cleanliness (optional)
x = np.arange(12).reshape(3, 4)
print(x)
```

We recreate x to remove the 999 change, so later examples start from clean data.

```python
# Negative indexing
x = np.arange(12).reshape(3, 4)

print("Last row:\n", x[-1])
print("Last column:\n", x[:, -1])
print("Bottom-right element:", x[-1, -1])
```

Negative indices count from the end.

-1 → last element

-2 → second last

So:

x[-1] → last row

x[:, -1] → last column

x[-1, -1] → bottom-right element

This is useful when array size is unknown.

```python
# Step slicing
print("Every 2nd column:\n", x[:, ::2])
print("Reverse rows:\n", x[::-1])
```

Slicing syntax:
start : stop : step

Examples:

::2 → every 2nd element

[::-1] → reverse order

So:

x[:, ::2] → all rows, every 2nd column

x[::-1] → rows reversed

```python
# Full slice vs copy difference
a = np.arange(6)

view = a[:]          # slice → view
copy = a.copy()      # explicit copy

view[0] = 999

print("a:", a)
print("copy:", copy)
```

a[:] → full slice → view

.copy() → new memory → copy

Changing view affects a.
Changing copy does not.

This reinforces the view concept from the memory chapter.

```python
# Shape change after slicing
x = np.arange(12).reshape(3, 4)

row = x[0]
col = x[:, 0]

print("Row shape:", row.shape)
print("Column shape:", col.shape)
```

Even though both look “linear”:

Row → shape (4,)

Column → shape (3,)

They are 1D views extracted from a 2D array.

This helps prevent confusion later in broadcasting.


## Summary

1. Indexing selects a single element: x[row, col]
2. Slicing selects ranges: x[row_start:row_end, col_start:col_end]
3. Slicing uses stop excluded (half-open ranges)
4. Basic slicing usually returns a view, so changes can affect the original array
