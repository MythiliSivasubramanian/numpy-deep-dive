# NumPy 2D Indexing, Filtering & Conditional Updates


## Topics covered
- Accessing elements in 2D arrays
- Boolean filtering
- Fancy indexing with row/column pairs
- Conditional updates
- Clipping / outlier handling
- Range replacements using masks


## Setup — Create the predefined array


````python
import numpy as np

b = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("Predefined Array:\n", b)
````

## Accessing a specific element

```python
b[1,2]
```

### How indexing works

Format:
```
array[row, column]
```

So:

| Value | Position |
|------|-----------|
| 60 | (1,2) |


````python
print("Element at position (1,2):", b[1,2])
````

### Visualizing indices


````python
for r in range(b.shape[0]):
    for c in range(b.shape[1]):
        print(f"Value {b[r,c]} at ({r},{c})")
````

##  Filter all values > 40

Boolean masking works the same in 2D arrays.

```python
b[b > 40]
```


````python
filtered = b[b > 40]

print("Values > 40:", filtered)
````

Notice the result is flattened into 1D — because we extracted matching values only.


## Fancy Indexing — Picking specific elements


We want:

| Value | Position |
|------|-----------|
| 10 | (0,0) |
| 60 | (1,2) |
| 80 | (2,1) |

Fancy indexing format:

```python
array[[rows],[columns]]
```


````python
picked = b[[0,1,2], [0,2,1]]

print("Picked elements (10, 60, 80):", picked)
````

### How it pairs indices

It pairs element‑wise:

```
(0,0)
(1,2)
(2,1)
```
Not a grid — but coordinate pairs.


## Update values > 50 → 999

Boolean masks can update values directly.


````python
b1 = b.copy()

b1[b1 > 50] = 999

print("Update >50 → 999:\n", b1)
````

## Update values < 30 → 0


````python
b2 = b.copy()

b2[b2 < 30] = 0

print("Update <30 → 0:\n", b2)
````

## Conditional replacement (manual clipping)


- Values > 80 → 80  
- Values < 20 → 20


````python
b3 = b.copy()

b3[b3 > 80] = 80
b3[b3 < 20] = 20

print("Manual clipping 20..80:\n", b3)
````

## Built‑in clipping — `np.clip()`

Cleaner, faster, production‑ready way.


````python
b4 = np.clip(b3, 20, 80)

print("np.clip result:\n", b4)
````

## Replace values in a range

```python
b[(b > 30) & (b < 70)] = 999
```


````python
b5 = b.copy()

b5[(b5 > 30) & (b5 < 70)] = 999

print("Values between 30 and 70 → 999:\n", b5)
````

## Mask breakdown 

Let’s inspect the mask itself.


````python
mask = (b > 30) & (b < 70)

print("Mask:\n", mask)
print("Filtered values:", b[mask])
````
