# NumPy Boolean Indexing (Masks): Range, AND/OR, NOT

- Select values in a **range** like `[3..8]`
- Combine conditions with **AND** (`&`) and **OR** (`|`)
- Invert a condition with **NOT** (`~mask`)
- Avoid common mistakes (very important ✅)

> **Key idea:** In NumPy, we often build a *boolean mask* (an array of `True/False`) and then use it to filter values.


## 1) Setup

Importing numpy

````python
import numpy as np

a = np.array([3, 8, 2, 9, 5, 1, 12, 7, 6, 4, 10])
a
````

## 2) Boolean masks: what they are

A **mask** is an array of booleans (`True/False`) created by comparing `a` with a condition.

Example: `a >= 5` produces a mask of the same length as `a`.

Then `a[mask]` returns **only** the elements where mask is `True`.


````python
mask_ge_5 = (a >= 5)
mask_ge_5
````

````python
a[mask_ge_5]
````

### Quick check (pairing values with True/False)

This is a useful way to *see* what the mask means.


````python
list(zip(a.tolist(), mask_ge_5.tolist()))
````

## 3) Scenario 1 — Range `[3..8]`


```python
between_3_8 = a[(a >= 3) & (a <= 8)]
```

### Explanation
- `(a >= 3)` is a boolean mask
- `(a <= 8)` is another boolean mask
- `&` combines them with **AND**
- Parentheses are required ✅


````python
between_3_8 = a[(a >= 3) & (a <= 8)]
between_3_8
````

### See the intermediate masks
Let's print each part to understand it deeply.


````python
mask_ge_3 = (a >= 3)
mask_le_8 = (a <= 8)
mask_range = mask_ge_3 & mask_le_8

print("a        :", a)
print("a >= 3   :", mask_ge_3)
print("a <= 8   :", mask_le_8)
print("AND (&)  :", mask_range)
print("filtered :", a[mask_range])
````

### More examples (practice)

Try different ranges:
- `[1..4]`
- `[6..12]`
- `[0..2]`


````python
def in_range(x, low, high):
    mask = (x >= low) & (x <= high)
    return x[mask]

print("range [1..4] :", in_range(a, 1, 4))
print("range [6..12]:", in_range(a, 6, 12))
print("range [0..2] :", in_range(a, 0, 2))
````

## 4) Scenario 2 — AND / OR conditions



```python
gt_5_and_even = a[(a > 5) & (a % 2 == 0)]
lt_4_or_gt_10 = a[(a < 4) | (a > 10)]
```

### What each part means
- `a > 5` → values greater than 5
- `a % 2 == 0` → even numbers (`%` is modulus)
- `&` → AND (both conditions must be True)
- `|` → OR (either condition can be True)


````python
gt_5_and_even = a[(a > 5) & (a % 2 == 0)]
lt_4_or_gt_10 = a[(a < 4) | (a > 10)]

print("a:", a)
print(">5 and even:", gt_5_and_even)
print("<4 or >10:", lt_4_or_gt_10)
````

### Break it down: `>5 AND even`

We compute masks step-by-step.


````python
mask_gt_5 = (a > 5)
mask_even = (a % 2 == 0)
mask_both = mask_gt_5 & mask_even

print("a         :", a)
print("a > 5     :", mask_gt_5)
print("a even    :", mask_even)
print("AND (&)   :", mask_both)
print("filtered  :", a[mask_both])
````

### Break it down: `<4 OR >10`

Again, mask-by-mask.


````python
mask_lt_4 = (a < 4)
mask_gt_10 = (a > 10)
mask_either = mask_lt_4 | mask_gt_10

print("a         :", a)
print("a < 4     :", mask_lt_4)
print("a > 10    :", mask_gt_10)
print("OR (|)    :", mask_either)
print("filtered  :", a[mask_either])
````

### More examples (practice)

1) Numbers that are **odd** OR **equal to 10**  
2) Numbers that are **between 4 and 9** AND **not equal to 7**  
3) Numbers that are **multiple of 3** OR **multiple of 5**


````python
# 1) odd OR equal to 10
ex1 = a[(a % 2 == 1) | (a == 10)]

# 2) between 4..9 AND not 7
ex2 = a[(a >= 4) & (a <= 9) & (a != 7)]

# 3) multiple of 3 OR multiple of 5
ex3 = a[(a % 3 == 0) | (a % 5 == 0)]

print("1) odd OR ==10:", ex1)
print("2) 4..9 AND !=7:", ex2)
print("3) mult of 3 OR 5:", ex3)
````

## 5) Scenario 3 — NOT with `~mask` (invert)

From your code:

```python
mask = (a >= 3) & (a <= 10)
outside = a[~mask]
```

### Explanation
- `mask` selects values **inside** `[3..10]`
- `~mask` flips `True ↔ False`
- `a[~mask]` selects values **outside** `[3..10]`


````python
mask_3_10 = (a >= 3) & (a <= 10)
outside = a[~mask_3_10]

print("a:", a)
print("mask (3..10):", mask_3_10)
print("inside:", a[mask_3_10])
print("outside:", outside)
````

### More NOT examples

- NOT even → odd numbers
- NOT in range `[6..9]` → everything except 6..9


````python
# NOT even => odd
odd = a[~(a % 2 == 0)]

# NOT in range 6..9
not_6_9 = a[~((a >= 6) & (a <= 9))]

print("odd:", odd)
print("not in [6..9]:", not_6_9)
````

## 6) Common mistakes 

### Mistake 1: Using Python `and` / `or` with NumPy arrays
This fails because `and/or` expect single True/False, not an array.
- Use `&` and `|` instead.

### Mistake 2: Forgetting parentheses
`a > 5 & a < 10` is parsed incorrectly.
- Write `(a > 5) & (a < 10)`.

### Mistake 3: Confusing `~` with negative sign
`~mask` is **bitwise NOT** on booleans, not numeric negation.
- Use `~` only on boolean masks.


````python
# Demonstrate correct parentheses usage:
correct = a[(a > 5) & (a < 10)]
correct
````

````python
import numpy as np

def scenario_range(a: np.ndarray) -> None:
    between_3_8 = a[(a >= 3) & (a <= 8)]
    print("Scenario: range [3..8]")
    print("a:", a)
    print("between:", between_3_8)
    print()

def scenario_and_or(a: np.ndarray) -> None:
    gt_5_and_even = a[(a > 5) & (a % 2 == 0)]
    lt_4_or_gt_10 = a[(a < 4) | (a > 10)]
    print("Scenario: AND / OR")
    print("a:", a)
    print(">5 and even:", gt_5_and_even)
    print("<4 or >10:", lt_4_or_gt_10)
    print()

def scenario_not(a: np.ndarray) -> None:
    mask = (a >= 3) & (a <= 10)
    outside = a[~mask]
    print("Scenario: NOT (~mask)")
    print("a:", a)
    print("mask (3..10):", mask)
    print("outside:", outside)
    print()

def main() -> None:
    a = np.array([3, 8, 2, 9, 5, 1, 12, 7, 6, 4, 10])
    scenario_range(a)
    scenario_and_or(a)
    scenario_not(a)

main()
````
