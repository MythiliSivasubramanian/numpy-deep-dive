# NumPy Array Conversion & Copying

This notebook explains how to:

- Convert Python lists ↔ NumPy arrays
- Understand views vs copies
- Avoid accidental data modification
- Use `.copy()` safely when working with slices

This concept is critical in Data Science workflows.


```python
import numpy as np
```

## 1) List → NumPy Array

Use `np.array()` to convert Python lists into NumPy arrays.


```python
lst = [1, 2, 3]
arr = np.array(lst)

print("lst:", lst)
print("arr:", arr)
print("dtype:", arr.dtype)
```

## 2) NumPy Array → List

Use `.tolist()` to convert arrays back to Python lists.


```python
back_to_list = arr.tolist()

print("Converted list:", back_to_list)
print("Type:", type(back_to_list))
```

## 3) View vs Copy

Slicing creates a **view** (shared memory).

Use `.copy()` to create an independent array.


```python
x = np.arange(6)

view = x[:]
copy = x.copy()

print("Before modification")
print("x   :", x)
print("view:", view)
print("copy:", copy)

view[0] = 999

print("\nAfter modifying view")
print("x   :", x)
print("view:", view)
print("copy:", copy)
```

## 4) Copy from Slice

Always use `.copy()` when you need independence from the source array.


```python
y = np.arange(10)

sub_copy = y[2:6].copy()
sub_copy[0] = -1

print("y       :", y)
print("sub_copy:", sub_copy)
```

## 5) Memory Sharing Check

Use `np.shares_memory()` to verify whether arrays share memory.


```python
x = np.arange(5)

view = x[:]
copy = x.copy()

print("View shares memory:", np.shares_memory(x, view))
print("Copy shares memory:", np.shares_memory(x, copy))
```

## Summary

- List ↔ Array conversion
- View vs Copy differences
- Safe slicing with `.copy()`
- Memory sharing checks
