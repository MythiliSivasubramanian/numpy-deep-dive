# NumPy Dimensions: 0D, 1D, 2D, 3D — `shape`, `ndim`, `size`

**array dimensions** in NumPy:

- What 0D, 1D, 2D, 3D arrays look like
- How to check dimensions using `shape` and `ndim`
- How to count total elements using `size`
- A key rule: **`ndim == len(shape)`**


```python
import numpy as np

def show_dim_info(name, arr):
    print(f"\n{name}")
    print(arr)
    print("shape:", arr.shape)
    print("ndim :", arr.ndim)
    print("size :", arr.size)
    print("dtype:", arr.dtype)
```

## 1) Quick Concepts

### `ndim`
- Number of dimensions (axes)
- 0D → scalar-like
- 1D → vector
- 2D → matrix/table (rows × columns)
- 3D → stack of matrices (depth × rows × columns)

### `shape`
- A tuple describing the **length of each axis**
- Examples:
  - 1D: `(4,)` → 4 elements
  - 2D: `(2, 3)` → 2 rows, 3 columns
  - 3D: `(2, 3, 4)` → 2 blocks, each with 3 rows, 4 cols

### `size`
- Total number of elements
- `size = shape[0] * shape[1] * ...`


## 2) Create 0D, 1D, 2D, 3D arrays

We will create one array of each dimension and inspect `shape`, `ndim`, `size`, and `dtype`.


```python
# 0D array (scalar-like)
a0 = np.array(5)

# 1D array (vector)
a1 = np.array([10, 20, 30, 40])

# 2D array (table/matrix)
a2 = np.array([[1, 2, 3],
               [4, 5, 6]])

# 3D array (stack of tables)
a3 = np.zeros((2, 3, 4))

show_dim_info("a0 (0D array)", a0)
show_dim_info("a1 (1D array)", a1)
show_dim_info("a2 (2D array)", a2)
show_dim_info("a3 (3D array)", a3)
```

## 3) How to interpret `shape` 

### Read `shape` left-to-right (axis by axis)

For a 2D array with `shape = (rows, cols)`:
- First number = number of **rows**
- Second number = number of **columns**

For a 3D array with `shape = (depth, rows, cols)`:
- First number = number of **blocks**
- Second number = rows per block
- Third number = columns per block


```python
print("a2.shape:", a2.shape, "-> rows =", a2.shape[0], ", cols =", a2.shape[1])
print("a3.shape:", a3.shape, "-> depth =", a3.shape[0], ", rows =", a3.shape[1], ", cols =", a3.shape[2])

print("\nTotal elements (size):")
print("a2.size should be 2*3 =", 2*3, "| actual:", a2.size)
print("a3.size should be 2*3*4 =", 2*3*4, "| actual:", a3.size)
```

## 4) Key Rule: `ndim == len(shape)`

This rule is almost always true in NumPy:

- `ndim` = number of axes
- `shape` = length of each axis
- Therefore, the number of axes = length of `shape`

Let's verify this for our arrays.


```python
print("Rule check: ndim == len(shape)")
for name, arr in [("a0", a0), ("a1", a1), ("a2", a2), ("a3", a3)]:
    print(name, "->", arr.ndim, "==", len(arr.shape), "?", arr.ndim == len(arr.shape))
```

## 5) Useful small operations

### Get length along each axis
- For 1D: `a.shape[0]`
- For 2D: `rows = a.shape[0]`, `cols = a.shape[1]`
- For 3D: `depth = a.shape[0]`, `rows = a.shape[1]`, `cols = a.shape[2]`

### Flatten size check
You can always compare `size` with a manual multiplication of the shape values.


```python
# Example: manual size calculation
def manual_size(arr):
    prod = 1
    for d in arr.shape:
        prod *= d
    return prod

for name, arr in [("a0", a0), ("a1", a1), ("a2", a2), ("a3", a3)]:
    print(name, "shape:", arr.shape, "manual_size:", manual_size(arr), "size:", arr.size)
```
