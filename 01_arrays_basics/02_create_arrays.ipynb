# NumPy Arrays – Creation & Basics

Below are the absolute fundamentals of **NumPy arrays**:

- Creating arrays (`np.array`, `np.zeros`, `np.ones`, `np.empty`, `np.arange`, `np.linspace`, `np.eye`, `np.diag`)
- Inspecting array properties: `shape`, `ndim`, `dtype`
- Understanding **Python list vs NumPy array** behavior

> **shape** and **dtype** —  are critical in Data Science and Machine Learning workflows.


```python
import numpy as np

def show_info(name, arr):
    print(f"\n{name}")
    print(arr)
    print("shape:", arr.shape)
    print("ndim :", arr.ndim)
    print("dtype:", arr.dtype)
```

## 1) Python List vs NumPy Array

Why do we need NumPy?

- Python lists are flexible but do not perform **element‑wise math** automatically.
- NumPy arrays are optimized for numerical operations.

Let’s compare the behavior of `* 2`.


```python
lst = [1, 2, 3]
arr = np.array(lst)

print("List vs NumPy array")
print("lst * 2 ->", lst * 2)   # list repeats
print("arr * 2 ->", arr * 2)   # element‑wise multiplication
```

## 2) Array Properties: `shape`, `ndim`, `dtype`

Three key attributes:

- **shape** → Structure of the array (rows, columns)
- **ndim** → Number of dimensions
- **dtype** → Data type of elements

These determine indexing, memory usage, and computations.


## 3) Creating Arrays

Below are the most commonly used NumPy array creation functions.


```python
# Create arrays
a1 = np.array([1, 2, 3])                   # 1D
a2 = np.array([[1, 2, 3], [4, 5, 6]])      # 2D

a3 = np.zeros((3, 4))                      # zeros # Type float
a4 = np.ones((2, 3))                       # ones # Type float
a5 = np.empty((2, 3))                      # uninitialized values
a6 = np.arange(0, 10, 2)                   # step-based range
a7 = np.linspace(0, 1, 5)                  # count-based range

a8 = np.eye(3)                             # identity matrix
a9 = np.diag([1, 2, 3])                    # diagonal matrix

a10 = np.zeros((2, 2), dtype=np.int32)     # dtype control example

show_info("a1 (np.array 1D)", a1)
show_info("a2 (np.array 2D)", a2)
show_info("a3 (np.zeros)", a3)
show_info("a4 (np.ones)", a4)
show_info("a5 (np.empty) NOTE: values not guaranteed", a5)
show_info("a6 (np.arange)", a6)
show_info("a7 (np.linspace)", a7)
show_info("a8 (np.eye)", a8)
show_info("a9 (np.diag)", a9)
show_info("a10 (zeros dtype=int32)", a10)
```

## 4) `np.zeros` and `np.ones`

Used for initialization (models, placeholders, accumulators).

Example task:
Create a `(4,2)` array filled with ones and inspect its properties.


```python
ex1 = np.ones((4, 2))
show_info("ex1 (ones 4x2)", ex1)
```

## 5) `np.empty` – Important Note

`np.empty(shape)` allocates memory but does **not initialize values**.

Values may look random because memory is reused.

Use only when you plan to overwrite values immediately.


```python
ex_empty = np.empty((2, 5))
show_info("ex_empty (empty 2x5)", ex_empty)

# Overwrite values
ex_empty[:] = 7
show_info("ex_empty after overwrite", ex_empty)
```

## 6) `np.arange` vs `np.linspace`

### `np.arange(start, stop, step)`
- Step-based
- `stop` usually excluded

### `np.linspace(start, stop, num)`
- Count-based
- Includes `stop`


```python
print("arange:", np.arange(0, 1, 0.2))
print("linspace:", np.linspace(0, 1, 6))
```

## 7) `np.eye` and `np.diag`

- `np.eye(n)` → Identity matrix
- `np.diag(v)` → Diagonal matrix from vector
- `np.diag(matrix)` → Extract diagonal


```python
I = np.eye(4)
show_info("Identity 4x4", I)

D = np.diag([9, 8, 7])
show_info("Diagonal matrix", D)

diag_vals = np.diag(D)
show_info("Extracted diagonal", diag_vals)
```

## 8) dtype – Memory & Performance

Controlling dtype impacts:

- Memory usage
- Speed
- ML compatibility


```python
x64 = np.zeros((2, 2), dtype=np.int64)
x32 = np.zeros((2, 2), dtype=np.int32)

show_info("int64 array", x64)
show_info("int32 array", x32)

print("Bytes per element:")
print("int64:", x64.itemsize)
print("int32:", x32.itemsize)
```
