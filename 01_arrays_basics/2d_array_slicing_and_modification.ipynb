## Problem
Create a **3×3 NumPy array** using values from **10 to 90** and perform multiple indexing and slicing operations.

## Objectives
1. Print array, `shape`, and `ndim`
2. Extract the **second row** → `[40, 50, 60]`
3. Extract the **third column** → `[30, 60, 90]`
4. Extract submatrix block:
   ```
   50 60
   80 90
   ```
5. Modify entire first column to `[999, 999, 999]`

## Concepts Covered
- 2D array creation using `arange()` and `reshape()`
- Row indexing
- Column slicing
- Submatrix extraction
- Column mutation (in-place update)


````
import numpy as np
````

## 1) Create a 3×3 Array (10 to 90)

We generate values using:

- `np.arange(10, 100, 10)` → 10, 20, ..., 90  
- `.reshape(3, 3)` → turns the 1D sequence into a 3×3 matrix


````
num_array = np.arange(10, 100, 10).reshape(3, 3)

print("2D Array:\n", num_array)
print("shape:", num_array.shape)
print("ndim :", num_array.ndim)
````

## 2) Extract the 2nd Row

In NumPy, rows are **0-indexed**:

- Row 0 → first row
- Row 1 → second row ✅
- Row 2 → third row

So the second row is:

```python
num_array[1]
```


````
second_row = num_array[1]
print("Second row:", second_row)
````

## 3) Extract the 3rd Column

To select a column:
- Use `:` to mean **all rows**
- Use the column index in the second position

The third column index is `2` (because 0, 1, 2).

```python
num_array[:, 2]
```


````
third_col = num_array[:, 2]
print("Third column:", third_col)
````

## 4) Extract a Submatrix Block

We want this block:

```
50 60
80 90
```

That corresponds to:
- Rows **1 to 2** (inclusive) → slice `1:3` (stop is exclusive)
- Columns **1 to 2** (inclusive) → slice `1:3`

```python
num_array[1:3, 1:3]
```


````
block = num_array[1:3, 1:3]
print("Extracted block:\n", block)
````

## 5) Modify the Entire First Column

To modify a full column:
- Use `:` for all rows
- Use column index `0` for the first column

```python
num_array[:, 0] = 999
```

This is an **in-place change**, meaning `num_array` itself is updated.


````
num_array[:, 0] = 999
print("Updated array:\n", num_array)
````

## Extra: Quick Shape Thinking

For a 2D array, you can always remember:

- `shape = (rows, cols)`
- `num_array[r, c]` accesses a single element
- `num_array[r, :]` selects a row
- `num_array[:, c]` selects a column
