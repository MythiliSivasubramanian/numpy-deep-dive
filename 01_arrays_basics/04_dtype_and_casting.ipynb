# NumPy dtypes & Casting — `dtype`, Inference, Conversion

In this notebook we focus on how NumPy handles **data types (dtypes)**:

Topics covered:

- Automatic dtype inference
- Why dtypes change (int → float upcasting)
- Mixed-type arrays
- Controlling dtype at creation
- Converting dtype using `astype()`
- Memory implications of dtype choices

> Understanding dtype is critical for performance, storage, and ML compatibility.


```python
import numpy as np

def show_dtype(name, arr):
    print(f"\n{name}")
    print(arr)
    print("dtype:", arr.dtype)
    print("shape:", arr.shape, "| ndim:", arr.ndim)
```

## 1) dtype Inference

NumPy automatically determines the dtype when you create arrays.

It inspects the values and chooses the most suitable common type.


```python
a = np.array([1, 2, 3])
b = np.array([1, 2, 3.0])   # contains float → upcasts
c = np.array([1, True, 3])  # mixed types → common dtype

show_dtype("a = np.array([1,2,3])", a)
show_dtype("b = np.array([1,2,3.0])", b)
show_dtype("c = np.array([1,True,3])", c)
```

## 2) Why dtype Changes (Upcasting)

NumPy promotes values to the **highest precision type** needed.

Examples:

- int + float → float
- int + bool → int
- int + string → string

This ensures no information is lost.


```python
x1 = np.array([1, 2, 3])
x2 = np.array([1.5, 2.5, 3.5])

result = x1 + x2
show_dtype("Result of int + float", result)
```

## 3) Controlling dtype at Creation

You can force NumPy to use a specific dtype.


```python
d = np.zeros((2, 2), dtype=np.int32)
show_dtype("d = zeros int32", d)

f = np.ones((2, 2), dtype=np.float32)
show_dtype("f = ones float32", f)
```

## 4) Converting dtype with `astype()`

`astype()` creates a **new array** with the requested dtype.

The original array remains unchanged.


```python
b = np.array([1, 2, 3.0])

e = b.astype(np.int32)

show_dtype("Original b (float)", b)
show_dtype("Converted e = b.astype(int32)", e)
```

## 5) Memory Impact of dtype

Different dtypes use different memory sizes.


```python
arr64 = np.zeros(5, dtype=np.int64)
arr32 = np.zeros(5, dtype=np.int32)

print("int64 itemsize:", arr64.itemsize, "bytes")
print("int32 itemsize:", arr32.itemsize, "bytes")
```

Try before checking solutions:

2. Convert it to int.
3. Create a `(3,3)` array of ones as float32.
4. Add an int array + float array and inspect dtype.


```python
# Solutions

ex1 = np.array([1,2,3,4], dtype=np.float64)
show_dtype("ex1 float", ex1)

ex2 = ex1.astype(np.int32)
show_dtype("ex2 converted to int32", ex2)

ex3 = np.ones((3,3), dtype=np.float32)
show_dtype("ex3 float32 ones", ex3)

ex4_a = np.array([1,2,3])
ex4_b = np.array([1.5,2.5,3.5])
ex4_res = ex4_a + ex4_b
show_dtype("ex4 result dtype", ex4_res)
```
