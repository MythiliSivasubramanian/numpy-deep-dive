# NumPy Array Attributes & Inspection

In this notebook we explore the most important **ndarray attributes** used to inspect arrays.

Focus areas:

- `shape` → structure of the array
- `ndim` → number of dimensions
- `size` → total number of elements
- `dtype` → data type
- `itemsize` → bytes per element
- `nbytes` → total memory usage

> These attributes helps to understand array structure, performance, and memory consumption.


````python
import numpy as np

def show_attrs(name, arr):
    print(f"\n{name}")
    print(arr)
    print("shape   :", arr.shape)
    print("ndim    :", arr.ndim)
    print("size    :", arr.size)
    print("dtype   :", arr.dtype)
    print("itemsize:", arr.itemsize)
    print("nbytes  :", arr.nbytes)
````

## 1) Understanding Key Attributes

### `shape`
Returns the structure of the array.

Examples:
- `(4,)` → 1D array with 4 elements
- `(3, 4)` → 2D array (3 rows, 4 columns)

### `ndim`
Number of axes/dimensions.

### `size`
Total number of elements (not rows or columns — all elements).

### `dtype`
Specifies the data type stored.

### `itemsize`
Memory used by one element (in bytes).

### `nbytes`
Total memory used by the array.


## 2) Create Sample Arrays

We will inspect attributes on arrays of different dimensions and dtypes.


````python
a1 = np.array([1, 2, 3, 4])
a2 = np.arange(12).reshape(3, 4)
a3 = np.zeros((2, 3), dtype=np.int32)

show_attrs("a1 (1D array)", a1)
show_attrs("a2 (2D array)", a2)
show_attrs("a3 (zeros int32)", a3)
````

## 3) Memory Understanding

Two important attributes:

- `itemsize` → bytes per element
- `nbytes` → total bytes used

Formula:

```
nbytes = size × itemsize
```


````python
print("Check: nbytes == size * itemsize")

for name, arr in [("a1", a1), ("a2", a2), ("a3", a3)]:
    print(name, "->", arr.nbytes == arr.size * arr.itemsize)
````

## 4) Compare Memory Across dtypes

Different dtypes use different memory sizes.


````python
x_int64 = np.zeros((5, 5), dtype=np.int64)
x_int32 = np.zeros((5, 5), dtype=np.int32)

show_attrs("x_int64", x_int64)
show_attrs("x_int32", x_int32)
````

## Summary

- Structure → `shape`
- Dimensions → `ndim`
- Element count → `size`
- Data type → `dtype`
- Memory per element → `itemsize`
- Total memory → `nbytes`

These attributes are essential for debugging, optimization, and ML dataset preparation.
